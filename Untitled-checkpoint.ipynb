```python
COUNTY_PREFIX = {"Northgate": "700", "Southfield": "880"}

PRECINCT_MERGE = {("Northgate", 253): 220, ("Southfield", 264): 30}

PARTY_MAP = {"DEM": "Dem", "REP": "Rep", "GRN": "Grn", "LIB": "Lib",
             "U": "Unaffiliated", "NPA": "Unaffiliated",
             "IND": "Unaffiliated", "Independent": "Unaffiliated"}


def make_precinct_key(county, precinct):
    precinct = PRECINCT_MERGE.get((county, int(precinct)), int(precinct))
    return COUNTY_PREFIX[county] + f"{precinct:03d}"


def clean_subdivision_name(s):
    if pd.isna(s):
        return None
    s = re.sub(r"\s*\([^)]*\)", "", str(s))                   
    s = re.sub(r"\b(PHASE|PH|SECTION|SEC|UNIT)\b.*$", "", s)   
    s = re.sub(r"^THE\s+", "", s.strip())                  
    s = s.strip().title()
    return s or None
```

```python
registered_voters = voters.drop_duplicates("SOS_VoterID").copy()

registered_voters["city_n"]   = registered_voters.city.str.strip().str.upper()
registered_voters["county_n"] = registered_voters.county.str.strip()
registered_voters["party_n"]  = registered_voters.party.map(PARTY_MAP).fillna("Unaffiliated")
registered_voters["sex_n"]    = registered_voters.sex.map({"M": "Male", "F": "Female"}).fillna("Unknown")
registered_voters.loc[(registered_voters.age < 18) | (registered_voters.age > 110), "age"] = np.nan
registered_voters["pctkey"]   = registered_voters.apply(lambda r: make_precinct_key(r.county_n, r.precinct_raw), axis=1)
registered_voters["subdiv_n"] = registered_voters.subdivision.apply(clean_subdivision_name)
registered_voters["is_lakeside"] = registered_voters.city_n == "LAKESIDE"

print("rows after dropping dups:", len(registered_voters))
print("eligible (Lakeside):", int(registered_voters.is_lakeside.sum()))
```

```python
northgate_inperson = northgate_raw.rename(columns={"CertificateNumber": "vid", "VoteDate": "date"})
northgate_inperson["method"] = "In-person"
northgate_inperson["date"]   = pd.to_datetime(northgate_inperson["date"], format="mixed").dt.date

southfield_inperson = southfield_raw.rename(columns={"voter_guid": "vid", "date_voted": "date"})
southfield_inperson["method"] = "In-person"
southfield_inperson["date"]   = pd.to_datetime(southfield_inperson["date"]).dt.date

mail_ballots = mail_raw.rename(columns={"VUID": "vid", "Return Date": "date"})
mail_ballots["method"] = "Mail"
mail_ballots["date"]   = pd.to_datetime(mail_ballots["date"]).dt.date

runoff_voters = pd.concat([northgate_inperson[["vid", "method", "date"]],
                           southfield_inperson[["vid", "method", "date"]],
                           mail_ballots[["vid", "method", "date"]]], ignore_index=True)

# in-person before mail, so drop_duplicates keeps the in-person row
runoff_voters["mrank"] = runoff_voters.method.map({"In-person": 0, "Mail": 1})
runoff_voters = (runoff_voters.sort_values(["mrank", "date"])
                              .drop_duplicates("vid")
                              .drop(columns="mrank")
                              .reset_index(drop=True))

print("distinct runoff voters:", len(runoff_voters), runoff_voters.method.value_counts().to_dict())
```

```python
first_round_may = first_round_raw.rename(columns={"voter_id": "vid"})
first_round_ids  = set(first_round_may.vid.unique())
runoff_voter_ids = set(runoff_voters.vid)

lakeside_voters = registered_voters[registered_voters.is_lakeside].copy()
lakeside_voters["voted_first"]  = lakeside_voters.SOS_VoterID.isin(first_round_ids)
lakeside_voters["voted_runoff"] = lakeside_voters.SOS_VoterID.isin(runoff_voter_ids)
lakeside_voters = lakeside_voters.merge(runoff_voters[["vid", "method", "date"]],
                                        left_on="SOS_VoterID", right_on="vid", how="left")

precinct_turnout = (lakeside_voters.groupby("pctkey")
                    .agg(reg=("SOS_VoterID", "size"),
                         first=("voted_first", "sum"),
                         runoff=("voted_runoff", "sum"))
                    .reset_index())
precinct_turnout["turnout"] = precinct_turnout.runoff / precinct_turnout.reg * 100
precinct_turnout["rel"]     = np.where(precinct_turnout["first"] > 0, precinct_turnout.runoff / precinct_turnout["first"] * 100, np.nan)
precinct_turnout["pnum"]    = precinct_turnout.pctkey.str[3:].astype(int)

subdivision_turnout = (lakeside_voters.dropna(subset=["subdiv_n"])
                       .groupby("subdiv_n")
                       .agg(reg=("SOS_VoterID", "size"),
                            first=("voted_first", "sum"),
                            runoff=("voted_runoff", "sum"))
                       .reset_index())
subdivision_turnout["turnout"] = subdivision_turnout.runoff / subdivision_turnout.reg * 100
subdivision_turnout["rel"]     = np.where(subdivision_turnout["first"] > 0, subdivision_turnout.runoff / subdivision_turnout["first"] * 100, np.nan)

precinct_turnout.sort_values("turnout", ascending=False)
```

```python
summary = dict(
    reg       = len(lakeside_voters),
    first     = int(lakeside_voters.voted_first.sum()),
    runoff    = int(lakeside_voters.voted_runoff.sum()),
    both      = len(runoff_voter_ids & first_round_ids),
    crossover = len(runoff_voter_ids - first_round_ids),                        
    yet       = len((first_round_ids & set(lakeside_voters.SOS_VoterID)) - runoff_voter_ids)) 
summary["turnout_pct"] = summary["runoff"] / summary["reg"] * 100
summary["rel_pct"]     = summary["runoff"] / summary["first"] * 100

for k, v in summary.items():
    print(f"  {k:12s}: {v:,.1f}" if isinstance(v, float) else f"  {k:12s}: {v:,}")
```

```python
for f in ["fonts/PlayfairDisplay.ttf", "fonts/Inter.ttf"]:
    try:
        fm.fontManager.addfont(f)
    except Exception:
        pass
SERIF = "Playfair Display" if any("Playfair" in f.name for f in fm.fontManager.ttflist) else "DejaVu Serif"
SANS = "Inter" if any(f.name == "Inter" for f in fm.fontManager.ttflist) else "DejaVu Sans"
plt.rcParams["font.family"] = SANS

BG = "#ECE7DB"; CARD = "#FBFAF5"; INK = "#1B1B19"; MUTE = "#7A756B"
LINE = "#D9D3C5"; TEAL = "#2E7D5B"; ORANGE = "#B05A1E"; BLUE = "#2F5AA8"
RED = "#B23A2E"; REDBG = "#F7E7E4"
GREENS = LinearSegmentedColormap.from_list("lk", ["#EDF2E9", "#A9CBA6", "#5E9E76", "#2C6B4C", "#1C4733"])

results = dict(lake=lakeside_voters, pct=precinct_turnout, sub=subdivision_turnout, kpis=summary)
summary, precinct_turnout, subdivision_turnout = results["kpis"], results["pct"], results["sub"]
lakeside_voters = results["lake"]
precinct_map = gpd.read_file("precincts.geojson").merge(precinct_turnout, on="pctkey")


def card(fig, x, y, w, h, label, value, *, highlight=False):
    ax = fig.add_axes([x, y, w, h]); ax.set_axis_off(); ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ec = RED if highlight else LINE
    fc = REDBG if highlight else CARD
    ax.add_patch(FancyBboxPatch((0.01, 0.04), 0.98, 0.92, boxstyle="round,pad=0.01,rounding_size=0.04",
                                fc=fc, ec=ec, lw=1.4 if highlight else 1.0))
    ax.text(0.5, 0.66, label, ha="center", va="center", fontsize=8.3, color=MUTE,
            family=SANS, fontweight="bold")
    ax.text(0.5, 0.30, value, ha="center", va="center", fontsize=27, color=RED if highlight else INK,
            family=SERIF, fontweight="bold")


def donut(fig, x, y, w, h, parts, colors, center, title):
    ax = fig.add_axes([x, y, w, h])
    ax.pie([p[1] for p in parts], colors=colors, startangle=90, counterclock=False,
           wedgeprops=dict(width=0.34, edgecolor=BG, linewidth=1.4))
    ax.text(0, 0, center, ha="center", va="center", fontsize=15, family=SERIF, fontweight="bold", color=INK)
    ax.set_title(title, fontsize=8.3, color=MUTE, fontweight="bold", family=SANS, loc="left", pad=6)
    
    lx = x + w + 0.004
    la = fig.add_axes([lx, y, 0.11, h]); la.set_axis_off(); la.set_ylim(0, 1); la.set_xlim(0, 1)
    n = len(parts)
    for i, (lab, val) in enumerate(parts):
        yy = 0.62 - i * (0.5 / max(n, 1))
        la.add_patch(plt.Rectangle((0.0, yy), 0.13, 0.10, color=colors[i]))
        la.text(0.2, yy + 0.05, f"{lab}", va="center", fontsize=8, color=INK)
        la.text(1.0, yy + 0.05, f"{val:,}", va="center", ha="right", fontsize=8,
                color=INK, fontweight="bold")


def age_hist(fig, x, y, w, h):
    ax = fig.add_axes([x, y, w, h])
    ages = lakeside_voters.loc[lakeside_voters.voted_runoff, "age"].dropna()
    bins = np.arange(18, 95, 5)
    ax.hist(ages, bins=bins, color=BLUE, edgecolor=BG, linewidth=0.6)
    if len(ages):
        ax.axvline(ages.median(), color=RED, ls="--", lw=1.2)
    for s in ax.spines.values():
        s.set_visible(False)
    ax.set_yticks([]); ax.tick_params(length=0, labelsize=7.5, colors=MUTE)
    ax.set_xticks([20, 30, 40, 50, 60, 70, 80, 90])
    ax.set_title("AGE OF RUNOFF VOTERS", fontsize=8.3, color=MUTE, fontweight="bold", family=SANS, loc="left", pad=6)


def choropleth(fig, x, y, w, h, value_col, title, eyebrow, headline):
    vmax = precinct_map[value_col].max()
    norm = plt.Normalize(0, vmax)
    ax = fig.add_axes([x, y, w, h]); ax.set_axis_off()
    precinct_map.plot(ax=ax, column=value_col, cmap=GREENS, norm=norm, edgecolor="white", linewidth=2.2)
    for _, r in precinct_map.iterrows():
        c = r.geometry.representative_point()
        v = r[value_col]
        txtc = "white" if (v / vmax if vmax else 0) > 0.55 else INK
        lab = "—" if pd.isna(v) else f"{v:.0f}%"
        ax.annotate(f"#{r.pnum}\n{lab}", (c.x, c.y), ha="center", va="center",
                    fontsize=8.5, color=txtc, fontweight="bold", linespacing=1.25)
    ax.set_xlim(*precinct_map.total_bounds[[0, 2]]); ax.set_ylim(*precinct_map.total_bounds[[1, 3]])
    # titles above map (bottom-anchored so they sit above the plot)
    fig.text(x, y + h + 0.028, eyebrow, fontsize=8, color=MUTE, fontweight="bold", family=SANS, va="bottom")
    fig.text(x, y + h + 0.010, title, fontsize=15.5, color=INK, family=SERIF, fontweight="bold", va="bottom")
    fig.text(x + w - 0.005, y + h + 0.012, headline, fontsize=21, color=TEAL, family=SERIF,
             fontweight="bold", ha="right", va="bottom")
    # colorbar
    cax = fig.add_axes([x, y - 0.016, w * 0.55, 0.006])
    cb = matplotlib.colorbar.ColorbarBase(cax, cmap=GREENS, norm=norm, orientation="horizontal")
    cb.outline.set_visible(False); cax.tick_params(length=0, labelsize=7, colors=MUTE)
    cb.set_ticks([0, vmax]); cax.set_xticklabels(["0%", f"{vmax:.0f}%"])


def rank_table(fig, x, y, w, h, df, value_col, header, color, ascending):
    rows = df.sort_values(value_col, ascending=ascending).head(5)
    ax = fig.add_axes([x, y, w, h]); ax.set_axis_off(); ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.add_patch(plt.Rectangle((0, 0.86), 1, 0.14, color=color))
    ax.text(0.04, 0.93, header, color="white", fontsize=8, fontweight="bold", va="center", family=SANS)
    for i, (_, r) in enumerate(rows.iterrows()):
        yy = 0.74 - i * 0.165
        ax.add_patch(plt.Rectangle((0, yy - 0.02), 1, 0.15, color=CARD if i % 2 == 0 else "#F2EEE4"))
        ax.text(0.04, yy + 0.05, f"D#{r.pnum}" if r.pctkey.startswith("700") else f"S#{r.pnum}",
                fontsize=8.5, va="center", color=INK, fontweight="bold")
        val = "—" if pd.isna(r[value_col]) else f"{r[value_col]:.1f}%"
        ax.text(0.66, yy + 0.05, val, fontsize=8.5, va="center", ha="right", color=INK, fontweight="bold")
        ax.text(0.97, yy + 0.05, f"n={r.reg}", fontsize=7.5, va="center", ha="right", color=MUTE)


def barh(fig, x, y, w, h, df, value_col, fmt, color, eyebrow, title, subtitle):
    rows = df.sort_values(value_col, ascending=False).head(8)
    ax = fig.add_axes([x, y, w, h])
    yp = np.arange(len(rows))[::-1]
    ax.barh(yp, rows[value_col], color=color, height=0.62)
    for yi, (_, r) in zip(yp, rows.iterrows()):
        ax.text(r[value_col], yi, "  " + (fmt.format(r[value_col])), va="center", fontsize=8, fontweight="bold", color=INK)
    ax.set_yticks(yp); ax.set_yticklabels(rows.subdiv_n, fontsize=8.2, color=INK)
    ax.set_xticks([])
    for s in ax.spines.values():
        s.set_visible(False)
    ax.tick_params(length=0)
    ax.set_xlim(0, rows[value_col].max() * 1.22)
    fig.text(x, y + h + 0.030, eyebrow, fontsize=7.5, color=MUTE, fontweight="bold", family=SANS, va="bottom")
    fig.text(x, y + h + 0.016, title, fontsize=13, color=INK, family=SERIF, fontweight="bold", va="bottom")
    fig.text(x, y + h + 0.004, subtitle, fontsize=7.5, color=MUTE, family=SANS, va="bottom")

fig = plt.figure(figsize=(12.4, 17.6), dpi=150)
fig.patch.set_facecolor(BG)

fig.text(0.5, 0.972, "Lakeside Runoff", ha="center", fontsize=46, family=SERIF, fontweight="bold", color=INK)
fig.text(0.5, 0.951, "City Council  ·  Early Vote through June 4, 2026", ha="center",
         fontsize=14, family=SERIF, color=MUTE, style="italic")
# logo mark
la = fig.add_axes([0.065, 0.948, 0.04, 0.028]); la.set_axis_off()
la.add_patch(plt.Rectangle((0.05, 0.1), 0.4, 0.8, color=TEAL))
la.add_patch(plt.Rectangle((0.5, 0.1), 0.45, 0.8, color=BLUE))
la.add_patch(plt.Circle((0.5, 0.5), 0.22, color=BG))
fig.text(0.115, 0.966, "LAKESIDE", fontsize=10, fontweight="bold", color=INK, family=SANS)
fig.text(0.115, 0.953, "TURNOUT WATCH", fontsize=8, color=MUTE, family=SANS)
fig.add_artist(plt.Line2D([0.05, 0.95], [0.939, 0.939], color=INK, lw=2.2, transform=fig.transFigure))


cards = [("REGISTERED VOTERS", f"{summary['reg']:,}", False),
         ("VOTED FIRST ROUND", f"{summary['first']:,}", False),
         ("VOTED RUNOFF", f"{summary['runoff']:,}", False),
         ("NEW IN RUNOFF\n(SKIPPED MAY)", f"{summary['crossover']:,}", True),
         ("MAY VOTERS\nNOT BACK YET", f"{summary['yet']:,}", False)]
cw = 0.172; gap = (0.90 - 5 * cw) / 4; x0 = 0.05
for i, (lab, val, hl) in enumerate(cards):
    card(fig, x0 + i * (cw + gap), 0.876, cw, 0.052, lab, val, highlight=hl)


sb = fig.add_axes([0.05, 0.788, 0.90, 0.076]); sb.set_axis_off(); sb.set_xlim(0, 1); sb.set_ylim(0, 1)
sb.add_patch(FancyBboxPatch((0.003, 0.02), 0.994, 0.96, boxstyle="round,pad=0.005,rounding_size=0.03",
                            fc=CARD, ec=LINE, lw=1.0))
donut(fig, 0.075, 0.792, 0.066, 0.066,
      [("Voted", summary["runoff"]), ("Not yet", summary["reg"] - summary["runoff"])],
      [BLUE, "#D8D2C4"], f"{summary['turnout_pct']:.1f}%", "TURNOUT")
method_counts = runoff_voters.method.value_counts()   # runoff already = distinct runoff voters
donut(fig, 0.375, 0.792, 0.066, 0.066,
      [("In-person", int(method_counts.get("In-person", 0))), ("Mail", int(method_counts.get("Mail", 0)))],
      [TEAL, ORANGE], f"{summary['runoff']}", "HOW THEY VOTED")
age_hist(fig, 0.66, 0.798, 0.275, 0.052)


fig.text(0.05, 0.762, "B Y   P R E C I N C T", fontsize=10.5, color=MUTE, fontweight="bold", family=SANS)
fig.add_artist(plt.Line2D([0.18, 0.95], [0.7635, 0.7635], color=LINE, lw=1.0, transform=fig.transFigure))

choropleth(fig, 0.06, 0.495, 0.38, 0.20, "rel", "Runoff \u00f7 First Round",
           "TURNOUT VS FIRST ROUND", f"{summary['rel_pct']:.1f}%")
choropleth(fig, 0.57, 0.495, 0.38, 0.20, "turnout", "Total Turnout",
           "SHARE OF REGISTERED", f"{summary['turnout_pct']:.1f}%")

rank_table(fig, 0.06, 0.41, 0.18, 0.075, precinct_turnout, "rel", "TOP 5", TEAL, ascending=False)
rank_table(fig, 0.26, 0.41, 0.18, 0.075, precinct_turnout, "rel", "BOTTOM 5", ORANGE, ascending=True)
rank_table(fig, 0.57, 0.41, 0.18, 0.075, precinct_turnout, "turnout", "TOP 5", TEAL, ascending=False)
rank_table(fig, 0.77, 0.41, 0.18, 0.075, precinct_turnout, "turnout", "BOTTOM 5", ORANGE, ascending=True)

# BY SUBDIVISION
fig.text(0.05, 0.378, "B Y   S U B D I V I S I O N", fontsize=10.5, color=MUTE, fontweight="bold", family=SANS)
fig.add_artist(plt.Line2D([0.21, 0.95], [0.3795, 0.3795], color=LINE, lw=1.0, transform=fig.transFigure))

barh(fig, 0.10, 0.07, 0.21, 0.245, subdivision_turnout, "runoff", "{:.0f}", BLUE,
     "VOLUME", "Runoff Voters", "Count of ballots cast.")
barh(fig, 0.42, 0.07, 0.21, 0.245, subdivision_turnout, "rel", "{:.0f}%", TEAL,
     "RELATIVE TO FIRST ROUND", "Runoff \u00f7 First Round", "Engagement vs. May.")
barh(fig, 0.74, 0.07, 0.21, 0.245, subdivision_turnout, "turnout", "{:.1f}%", ORANGE,
     "SHARE OF REGISTERED", "Turnout", "Runoff \u00f7 registered.")

fig.add_artist(plt.Line2D([0.05, 0.95], [0.045, 0.045], color=LINE, lw=1.0, transform=fig.transFigure))
fig.text(0.5, 0.028, "Lakeside Turnout Watch  ·  Sources: Northgate & Southfield county early-vote rosters, "
         "registered-voter roll, May first-round file  ·  Through June 4, 2026",
         ha="center", fontsize=8, color=MUTE, family=SANS)

import os
os.makedirs("outputs", exist_ok=True)
fig.savefig("outputs/lakeside_runoff_report.png", facecolor=BG, bbox_inches="tight", pad_inches=0.25)
print("saved outputs/lakeside_runoff_report.png")
```
